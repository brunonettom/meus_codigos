```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from scipy import stats

# Enable inline plotting for matplotlib
%matplotlib inline

# Set the style for seaborn plots
sns.set(style="whitegrid")

# Create the dataset from the provided values
data = [19.00000, 15.00000, 17.63333, 17.83333, 19.40000, 19.43333, 21.50000, 12.13333,
        12.15000, 12.20000, 17.80000, 21.01667, 21.05000, 11.26667, 11.08333, 11.08333,
        18.51667, 12.43333, 12.20000, 20.23333, 15.50000, 14.35000, 13.71667, 12.81667,
        11.96667, 17.21667, 10.58333, 13.71667, 10.58333, 20.00000, 19.91667, 13.21667,
        13.23333, 17.48333, 12.71667, 19.81667, 14.55000, 14.56667, 17.45000, 12.08333,
        8.58333, 11.61667, 11.60000, 12.63333, 15.76667]

# Create a pandas DataFrame
df = pd.DataFrame(data, columns=['Value'])

# Display the first few rows of the dataset
df.head()
```

# Análise Exploratória de Dados

Neste notebook, vamos analisar um conjunto de dados numéricos usando diversas técnicas de visualização e análise estatística. O objetivo é extrair insights importantes dos dados através de diferentes perspectivas.

```python
# Análise Estatística Básica

# Estatísticas descritivas
desc_stats = df.describe()
print("Estatísticas descritivas:")
print(desc_stats)

# Verificando se existem valores nulos
print("\nValores nulos:")
print(df.isnull().sum())

# Calculando medidas adicionais
print("\nMediana:", df['Value'].median())
print("Moda:", df['Value'].mode()[0])
print("Variância:", df['Value'].var())
print("Desvio padrão:", df['Value'].std())
print("Amplitude (Range):", df['Value'].max() - df['Value'].min())
print("Quartis:")
print(df['Value'].quantile([0.25, 0.5, 0.75]))
```

## Análise da Distribuição dos Dados

Vamos examinar a distribuição dos valores utilizando diferentes visualizações para entender melhor o comportamento dos dados.

```python
# Criando um histograma com curva de densidade (KDE)
plt.figure(figsize=(12, 6))
sns.histplot(df['Value'], kde=True, bins=10, color='skyblue')
plt.title('Histograma com Curva de Densidade', fontsize=15)
plt.xlabel('Valor', fontsize=12)
plt.ylabel('Frequência', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

# Criando um gráfico de densidade (KDE) separado para melhor visualização
plt.figure(figsize=(12, 6))
sns.kdeplot(df['Value'], fill=True, color='green', alpha=0.7)
plt.title('Curva de Densidade (KDE)', fontsize=15)
plt.xlabel('Valor', fontsize=12)
plt.ylabel('Densidade', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Box Plot para visualizar a distribuição e outliers
plt.figure(figsize=(12, 6))
sns.boxplot(x=df['Value'], color='skyblue')
plt.title('Box Plot dos Valores', fontsize=15)
plt.xlabel('Valor', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

# Violin Plot para visualizar a distribuição de forma mais detalhada
plt.figure(figsize=(12, 6))
sns.violinplot(x=df['Value'], color='lightgreen')
plt.title('Violin Plot dos Valores', fontsize=15)
plt.xlabel('Valor', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

# Combinando Box Plot e Strip Plot para ver pontos individuais
plt.figure(figsize=(12, 6))
sns.boxplot(x=df['Value'], color='skyblue')
sns.stripplot(x=df['Value'], color='red', alpha=0.6, jitter=True)
plt.title('Box Plot com Pontos Individuais', fontsize=15)
plt.xlabel('Valor', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

## Análise de Série Temporal

Embora não tenhamos informações temporais explícitas, vamos considerar a ordem dos dados como uma sequência temporal para visualizar tendências e padrões.

```python
# Adicionando uma coluna de índice como 'Tempo'
df['Tempo'] = range(1, len(df) + 1)

# Gráfico de linha simples
plt.figure(figsize=(14, 7))
plt.plot(df['Tempo'], df['Value'], marker='o', linestyle='-', color='blue', alpha=0.7)
plt.title('Série Temporal dos Valores', fontsize=15)
plt.xlabel('Sequência', fontsize=12)
plt.ylabel('Valor', fontsize=12)
plt.grid(True, alpha=0.3)
plt.xticks(df['Tempo'][::5])  # Mostrar apenas alguns pontos no eixo x para clareza
plt.show()

# Gráfico de linha com média móvel para visualizar tendências
plt.figure(figsize=(14, 7))
plt.plot(df['Tempo'], df['Value'], marker='o', linestyle='-', color='blue', alpha=0.5, label='Valor Original')

# Calculando média móvel com janela de 3 pontos
df['MA3'] = df['Value'].rolling(window=3).mean()
plt.plot(df['Tempo'], df['MA3'], linestyle='-', color='red', linewidth=2, label='Média Móvel (3 pontos)')

# Calculando média móvel com janela de 5 pontos
df['MA5'] = df['Value'].rolling(window=5).mean()
plt.plot(df['Tempo'], df['MA5'], linestyle='-', color='green', linewidth=2, label='Média Móvel (5 pontos)')

plt.title('Série Temporal com Médias Móveis', fontsize=15)
plt.xlabel('Sequência', fontsize=12)
plt.ylabel('Valor', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.xticks(df['Tempo'][::5])
plt.show()
```

## Agrupamento e Segmentação dos Dados

Podemos tentar agrupar os dados em diferentes categorias para identificar padrões.

```python
# Segmentação dos dados em faixas de valores

# Definindo os bins (categorias)
bins = [0, 10, 13, 16, 19, 22]
labels = ['Muito Baixo', 'Baixo', 'Médio', 'Alto', 'Muito Alto']

# Categorizando os valores
df['Categoria'] = pd.cut(df['Value'], bins=bins, labels=labels)

# Contagem por categoria
cat_counts = df['Categoria'].value_counts().sort_index()
print("Contagem por categoria:")
print(cat_counts)

# Visualizando a distribuição das categorias
plt.figure(figsize=(12, 6))
sns.countplot(x='Categoria', data=df, palette='viridis')
plt.title('Distribuição por Categoria', fontsize=15)
plt.xlabel('Categoria', fontsize=12)
plt.ylabel('Contagem', fontsize=12)
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)
plt.show()

# Criando um gráfico de barras interativo com Plotly
fig = px.bar(cat_counts.reset_index(), x='Categoria', y='count', 
             title='Distribuição por Categoria',
             labels={'count': 'Contagem', 'Categoria': 'Categoria'},
             color='Categoria',
             color_discrete_sequence=px.colors.qualitative.Set2)

fig.update_layout(
    title_font_size=20,
    xaxis_title_font_size=16,
    yaxis_title_font_size=16,
    width=900,
    height=500
)

fig.show()
```

## Detecção de Anomalias

Vamos identificar possíveis outliers e valores anômalos no conjunto de dados.

```python
# Detecção de outliers usando o método IQR (Intervalo Interquartil)
Q1 = df['Value'].quantile(0.25)
Q3 = df['Value'].quantile(0.75)
IQR = Q3 - Q1

# Definindo os limites para outliers
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print(f"Intervalo Interquartil (IQR): {IQR:.2f}")
print(f"Limite inferior para outliers: {limite_inferior:.2f}")
print(f"Limite superior para outliers: {limite_superior:.2f}")

# Identificando outliers
outliers = df[(df['Value'] < limite_inferior) | (df['Value'] > limite_superior)]
print(f"\nNúmero de outliers identificados: {len(outliers)}")

if len(outliers) > 0:
    print("\nOutliers:")
    print(outliers)

# Visualizando os outliers em um gráfico de dispersão
plt.figure(figsize=(14, 7))
sns.scatterplot(x='Tempo', y='Value', data=df, s=100, color='blue', alpha=0.6)

# Destacando os outliers em vermelho
if len(outliers) > 0:
    sns.scatterplot(x='Tempo', y='Value', data=outliers, s=150, color='red', marker='X', label='Outliers')

plt.axhline(y=limite_superior, color='red', linestyle='--', alpha=0.7, label=f'Limite Superior ({limite_superior:.2f})')
plt.axhline(y=limite_inferior, color='red', linestyle='--', alpha=0.7, label=f'Limite Inferior ({limite_inferior:.2f})')

plt.title('Identificação de Outliers pelo Método IQR', fontsize=15)
plt.xlabel('Sequência', fontsize=12)
plt.ylabel('Valor', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## Análise de Correlação e Autocorrelação

Vamos investigar se existe alguma correlação entre os valores consecutivos na série.

```python
# Calculando a autocorrelação
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# Criando o gráfico de autocorrelação
plt.figure(figsize=(12, 6))
plot_acf(df['Value'], lags=20, alpha=0.05)
plt.title('Função de Autocorrelação (ACF)', fontsize=15)
plt.grid(True, alpha=0.3)
plt.show()

# Criando o gráfico de autocorrelação parcial
plt.figure(figsize=(12, 6))
plot_pacf(df['Value'], lags=20, alpha=0.05)
plt.title('Função de Autocorrelação Parcial (PACF)', fontsize=15)
plt.grid(True, alpha=0.3)
plt.show()

# Comparando valores consecutivos
df['Valor_Anterior'] = df['Value'].shift(1)

# Removendo o primeiro valor (que não tem um valor anterior)
df_corr = df.dropna()

# Criando um gráfico de dispersão entre valores consecutivos
plt.figure(figsize=(10, 10))
sns.scatterplot(x='Valor_Anterior', y='Value', data=df_corr, s=100, alpha=0.7)
plt.title('Relação entre Valores Consecutivos', fontsize=15)
plt.xlabel('Valor Anterior', fontsize=12)
plt.ylabel('Valor Atual', fontsize=12)
plt.grid(True, alpha=0.3)

# Adicionando uma linha diagonal para referência
plt.plot([df_corr['Valor_Anterior'].min(), df_corr['Valor_Anterior'].max()], 
         [df_corr['Valor_Anterior'].min(), df_corr['Valor_Anterior'].max()], 
         'k--', alpha=0.5)

# Calculando e mostrando o coeficiente de correlação
corr = df_corr['Value'].corr(df_corr['Valor_Anterior'])
plt.annotate(f'Correlação: {corr:.2f}', 
             xy=(0.05, 0.95), xycoords='axes fraction', 
             fontsize=12, bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.8))

plt.show()
```

## Ajuste de Distribuições de Probabilidade

Vamos verificar se os dados seguem alguma distribuição de probabilidade conhecida.

```python
# Ajuste de distribuições teóricas
from scipy import stats
import matplotlib.pyplot as plt
import numpy as np

# Configurando a figura
plt.figure(figsize=(14, 10))

# Preparando os dados
x = np.linspace(min(df['Value']), max(df['Value']), 100)
values = df['Value']

# Testando várias distribuições
distributions = [
    (stats.norm, 'Normal'),
    (stats.expon, 'Exponencial'),
    (stats.gamma, 'Gamma'),
    (stats.lognorm, 'Log-Normal')
]

# Configurando o subplot
plt.subplot(2, 2, 1)
sns.histplot(values, kde=True, stat="density", alpha=0.4, color='skyblue')

# Testando cada distribuição e adicionando ao gráfico
for i, (distribution, name) in enumerate(distributions):
    # Ajustando a distribuição
    params = distribution.fit(values)
    
    # Calculando a função de densidade de probabilidade
    pdf = distribution.pdf(x, *params)
    
    # Adicionando ao gráfico
    plt.plot(x, pdf, label=f'Distribuição {name}')
    
    # Teste de Kolmogorov-Smirnov para verificar o ajuste
    ks_stat, p_value = stats.kstest(values, distribution.cdf, args=params)
    print(f"Teste KS para distribuição {name}: estatística = {ks_stat:.4f}, p-value = {p_value:.4f}")
    
    if i < 3:  # Para as próximas distribuições
        plt.subplot(2, 2, i+2)
        sns.histplot(values, kde=True, stat="density", alpha=0.4, color='skyblue')
        plt.plot(x, pdf, label=f'Distribuição {name}')
        plt.title(f'Ajuste da Distribuição {name}', fontsize=12)
        plt.xlabel('Valor', fontsize=10)
        plt.ylabel('Densidade', fontsize=10)
        plt.legend()
        plt.grid(True, alpha=0.3)

# Configurando o primeiro subplot
plt.subplot(2, 2, 1)
plt.title('Comparação de Distribuições Teóricas', fontsize=12)
plt.xlabel('Valor', fontsize=10)
plt.ylabel('Densidade', fontsize=10)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Transformações de Dados

Vamos explorar algumas transformações nos dados para verificar se alteram a distribuição ou revelam padrões.

```python
# Testando várias transformações

# Criando uma cópia dos dados para não alterar o DataFrame original
df_transform = df.copy()

# Transformação logarítmica
df_transform['Log'] = np.log(df['Value'])

# Transformação de raiz quadrada
df_transform['Sqrt'] = np.sqrt(df['Value'])

# Transformação quadrática
df_transform['Squared'] = df['Value'] ** 2

# Transformação recíproca (inversa)
df_transform['Reciprocal'] = 1 / df['Value']

# Visualizando as transformações
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Original
sns.histplot(df_transform['Value'], kde=True, ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Dados Originais', fontsize=12)
axes[0, 0].set_xlabel('Valor', fontsize=10)
axes[0, 0].set_ylabel('Frequência', fontsize=10)
axes[0, 0].grid(True, alpha=0.3)

# Log
sns.histplot(df_transform['Log'], kde=True, ax=axes[0, 1], color='green')
axes[0, 1].set_title('Transformação Logarítmica', fontsize=12)
axes[0, 1].set_xlabel('Log(Valor)', fontsize=10)
axes[0, 1].set_ylabel('Frequência', fontsize=10)
axes[0, 1].grid(True, alpha=0.3)

# Sqrt
sns.histplot(df_transform['Sqrt'], kde=True, ax=axes[1, 0], color='orange')
axes[1, 0].set_title('Transformação Raiz Quadrada', fontsize=12)
axes[1, 0].set_xlabel('Sqrt(Valor)', fontsize=10)
axes[1, 0].set_ylabel('Frequência', fontsize=10)
axes[1, 0].grid(True, alpha=0.3)

# Squared
sns.histplot(df_transform['Squared'], kde=True, ax=axes[1, 1], color='purple')
axes[1, 1].set_title('Transformação Quadrática', fontsize=12)
axes[1, 1].set_xlabel('Valor²', fontsize=10)
axes[1, 1].set_ylabel('Frequência', fontsize=10)
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Visualizando a transformação recíproca separadamente (pode ter escala muito diferente)
plt.figure(figsize=(12, 6))
sns.histplot(df_transform['Reciprocal'], kde=True, color='red')
plt.title('Transformação Recíproca (1/Valor)', fontsize=15)
plt.xlabel('1/Valor', fontsize=12)
plt.ylabel('Frequência', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```
